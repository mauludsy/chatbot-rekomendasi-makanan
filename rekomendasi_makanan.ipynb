MEMBERSIHKAN DATA DUPLIKAT

```python
import re

def clean_title(title):
    title = str(title).lower()
    title = re.sub(r'#kitaberbagi', '', title, flags=re.IGNORECASE)
    title = re.sub(r'[^a-zA-Z0-9\s]', '', title)
    title = re.sub(r'\s+', ' ', title).strip()
    return title
```

CLEANING DAN VALIDASI JUDUL MAKANAN

```python
rekomendasi_makanan['Title Cleaned'] = rekomendasi_makanan['Title'].apply(clean_title)

rekomendasi_makanan.loc[
    rekomendasi_makanan['Title Cleaned'] == '',
    'Title Cleaned'
] = rekomendasi_makanan['Title']
```

```python
import re

def clean_ingredients(text):
    text = str(text).lower()
    
    # Hapus URL
    text = re.sub(r'http\S+|www\S+', ' ', text)
    
    # Hapus underscore
    text = re.sub(r'_+', ' ', text)
    
    # Hapus angka
    text = re.sub(r'\d+', ' ', text)
    
    # Hapus karakter selain huruf dan spasi
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    
    # Hapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
```

```python
def detect_preferences(user_input):
    user_input = user_input.lower()
    
    detected = []
    
    for preference, keywords in preference_keywords.items():
        for keyword in keywords:
            if keyword in user_input:
                detected.append(preference)
                break
    
    return detected
```

```python
ingredient_vocabulary = set(
    tfidf.get_feature_names_out()
)
```

```python
import re

def detect_ingredients(user_input):
    user_input = user_input.lower()
    
    # Pecah input menjadi kata
    user_words = set(re.findall(r'\b[a-zA-Z]+\b', user_input))
    
    detected = []
    
    for ingredient in ingredient_vocabulary:
        if ingredient in user_words:
            detected.append(ingredient)
    
    return detected
```

```python
user_input = "Saya ingin makanan yang menggunakan ayam dan rasanya pedas"

ingredients = detect_ingredients(user_input)

print(ingredients)
```

```python
import re

def detect_ingredients(user_input):
    user_input = user_input.lower()
    
    detected = []
    
    for ingredient in common_ingredients:
        pattern = r'\b' + re.escape(ingredient) + r'\b'
        
        if re.search(pattern, user_input):
            detected.append(ingredient)
    
    return detected
```

```python
user_input = "Saya ingin makanan yang menggunakan ayam dan rasanya pedas"

ingredients = detect_ingredients(user_input)

print("Bahan:", ingredients)
```

```python
def filter_by_ingredients(df, ingredients):
    candidate_df = df.copy()
    
    for ingredient in ingredients:
        candidate_df = candidate_df[
            candidate_df['Ingredients Cleaned'].str.contains(
                r'\b' + re.escape(ingredient) + r'\b',
                case=False,
                na=False,
                regex=True
            )
        ]
    
    return candidate_df
```

```python
candidate_df = filter_by_ingredients(
    rekomendasi_makanan,
    ingredients
)

print("Jumlah resep awal :", len(rekomendasi_makanan))
print("Jumlah kandidat   :", len(candidate_df))
```

```python
preference_keywords = {
    'pedas': [
        'pedas',
        'cabe',
        'cabai',
        'sambal',
        'lada',
        'merica'
    ],
    
    'manis': [
        'manis',
        'gula',
        'madu',
        'kecap manis'
    ],
    
    'asam': [
        'asam',
        'jeruk',
        'cuka',
        'asam jawa'
    ],
    
    'gurih': [
        'gurih',
        'kaldu',
        'santan',
        'keju'
    ]
}
```

```python
def calculate_preference_score(ingredients_text, preferences):
    
    ingredients_text = ingredients_text.lower()
    
    score = 0
    
    for preference in preferences:
        
        keywords = preference_keywords.get(preference, [])
        
        for keyword in keywords:
            
            if re.search(
                r'\b' + re.escape(keyword) + r'\b',
                ingredients_text
            ):
                score += 1
    
    return score
```

```python
candidate_df['Preference Score'] = candidate_df[
    'Ingredients Cleaned'
].apply(
    lambda x: calculate_preference_score(
        x,
        preferences
    )
)
```

```python
# Index kandidat
candidate_indices = candidate_df.index

# Ambil matriks TF-IDF kandidat
candidate_matrix = tfidf_matrix[candidate_indices]

# TF-IDF input user
user_vector = tfidf.transform([user_input])

# Cosine similarity
similarity_scores = cosine_similarity(
    user_vector,
    candidate_matrix
).flatten()

# Masukkan similarity ke dataframe
candidate_df['Similarity'] = similarity_scores
```

Normalisasi Preference Score

```python
max_preference = candidate_df['Preference Score'].max()

if max_preference > 0:
    candidate_df['Preference Normalized'] = (
        candidate_df['Preference Score'] / max_preference
    )
else:
    candidate_df['Preference Normalized'] = 0
```

Buat Final Score

```python
candidate_df['Final Score'] = (
    candidate_df['Similarity'] * 0.7
    +
    candidate_df['Preference Normalized'] * 0.3
)
```

```python
import re
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity


def rekomendasi(user_input, top_n=5):
    
    # ==========================================
    # 1. Deteksi bahan
    # ==========================================
    ingredients = detect_ingredients(user_input)
    
    # ==========================================
    # 2. Deteksi preferensi
    # ==========================================
    preferences = detect_preferences(user_input)
    
    print("Input       :", user_input)
    print("Bahan       :", ingredients)
    print("Preferensi  :", preferences)
    
    # ==========================================
    # 3. Filter berdasarkan bahan
    # ==========================================
    if ingredients:
        candidate_df = filter_by_ingredients(
            rekomendasi_makanan,
            ingredients
        ).copy()
    else:
        candidate_df = rekomendasi_makanan.copy()
    
    # Jika tidak ada kandidat
    if len(candidate_df) == 0:
        print("\nTidak ditemukan resep yang sesuai.")
        return None
    
    # ==========================================
    # 4. TF-IDF Similarity
    # ==========================================
    candidate_indices = candidate_df.index
    
    candidate_matrix = tfidf_matrix[candidate_indices]
    
    user_vector = tfidf.transform([user_input])
    
    similarity_scores = cosine_similarity(
        user_vector,
        candidate_matrix
    ).flatten()
    
    candidate_df['Similarity'] = similarity_scores
    
    # ==========================================
    # 5. Preference Score
    # ==========================================
    candidate_df['Preference Score'] = candidate_df[
        'Ingredients Cleaned'
    ].apply(
        lambda x: calculate_preference_score(
            x,
            preferences
        )
    )
    
    # ==========================================
    # 6. Normalisasi Preference Score
    # ==========================================
    max_preference = candidate_df['Preference Score'].max()
    
    if max_preference > 0:
        candidate_df['Preference Normalized'] = (
            candidate_df['Preference Score'] /
            max_preference
        )
    else:
        candidate_df['Preference Normalized'] = 0
    
    # ==========================================
    # 7. Final Score
    # ==========================================
    candidate_df['Final Score'] = (
        candidate_df['Similarity'] * 0.7
        +
        candidate_df['Preference Normalized'] * 0.3
    )
    
    # ==========================================
    # 8. Ranking
    # ==========================================
    hasil = candidate_df.sort_values(
        'Final Score',
        ascending=False
    ).head(top_n)
    
    # ==========================================
    # 9. Tampilkan hasil
    # ==========================================
    print("\nRekomendasi makanan:")
    
    display(
        hasil[
            [
                'Title',
                'Category',
                'Similarity',
                'Preference Score',
                'Final Score'
            ]
        ].reset_index(drop=True)
    )
    
    return hasil
```
